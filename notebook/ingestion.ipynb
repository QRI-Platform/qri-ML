```python
from langchain_docling import DoclingLoader
from langchain_docling.loader import ExportType
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.datamodel.base_models import InputFormat

# 1. Pipeline Options configure karein
pipeline_options = PdfPipelineOptions()
pipeline_options.do_ocr = True
pipeline_options.do_formula_enrichment = True  # Equation decoding on

# 2. Custom DocumentConverter banayein
custom_converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options),
        InputFormat.IMAGE: PdfFormatOption(pipeline_options=pipeline_options),
    }
)

# 3. LangChain Loader me converter pass karein
loader = DoclingLoader(
    file_path="https://imgs.search.brave.com/zTsvkxUm4EZeipWuExgh4Ylhkbq8Oa_busoHAd7TuBE/rs:fit:860:0:0:0/g:ce/aHR0cHM6Ly9jbi5l/ZHVyZXYuaW4vQXBw/bGljYXRpb25JbWFn/ZXMvVGVtcC8xNDE5/Njc3X2ZhZGMxZmM0/LTVlYWMtNDA2YS1i/YjU5LWQ2NzVkNGYw/MzAwY19sZy5wbmc",
    converter=custom_converter,
    export_type=ExportType.MARKDOWN  # Output ko direct Markdown format me lane ke liye
)

# 4. Load LangChain Document objects
docs = loader.load()

# Extracted Markdown content
for doc in docs:
    print(doc.page_content)
```
